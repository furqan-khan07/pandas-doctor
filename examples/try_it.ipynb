Some data: 200,000 orders.

````python
import numpy as np
import pandas as pd

rng = np.random.default_rng(0)
n = 200_000
df = pd.DataFrame(
    {
        "price": rng.gamma(2.0, 10.0, n).round(2),
        "qty": rng.integers(1, 6, n),
        "country": rng.choice(["CA", "US", "UK"], n),
    }
)
df.head()
````

A loop a lot of notebooks have. `%%doctor` runs the cell as usual, then checks a vectorised rewrite on the same data and times both.

````python
%%doctor
df["revenue"] = 0.0
for i, row in df.iterrows():
    df.at[i, "revenue"] = row["price"] * row["qty"]
````

Row-wise `apply`. With `--rewrite`, the verified version is dropped into a new cell below.

````python
%%doctor --rewrite
df["big_order"] = df.apply(lambda r: r["price"] * r["qty"] > 100, axis=1)
````
